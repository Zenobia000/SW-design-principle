# 里氏替換原則 (Liskov Substitution Principle, LSP)
## 定義
子類別應該可以替代其父類別，並確保系統行為保持正確一致。

## 判斷導入時機

- Do：
當子類別能完全替代父類別，且不改變系統行為時。
需要確保多態性正確運作，避免潛在的運行時錯誤。

- Don't：
當子類別違反父類別的行為約定，導致功能不一致時。


----


## 完整案例

### 需求情境

系統需要管理多種類型的鳥類 (Bird)，要求所有鳥類都有移動功能 (move)。
不同的鳥可能會有不同的移動方式，例如：麻雀 (Sparrow) 可以飛行，而鴕鳥 (Ostrich) 只能奔跑。

### 初始設計


```python
class Bird:
    def fly(self):
        pass

class Sparrow(Bird):
    def fly(self):
        print("Sparrow is flying in the sky.")

class Ostrich(Bird):
    def fly(self):
        raise NotImplementedError("Ostriches cannot fly.")

# 測試代碼
birds = [Sparrow(), Ostrich()]

for bird in birds:
    bird.fly()
```

> 問題分析

1. Ostrich 繼承了 Bird，但不能實現 fly 的行為，導致客戶端代碼的預期行為被破壞。
2. 如果新增其他鳥類如企鵝 (Penguin)，需要進一步檢查或處理額外行為，會增加維護成本。

### 重構設計
解決方案：

將鳥類分為可飛行與不可飛行的子類別，避免單一父類別的行為不一致問題。
使用抽象基類 (Abstract Base Class, ABC) 定義行為，確保類別的正確替代。

```python
from abc import ABC, abstractmethod

# 抽象基類：所有鳥類都應該實現 move 方法
class Bird(ABC):
    @abstractmethod
    def move(self):
        pass

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def get_name(self):
        pass

# 可飛行的鳥類
class FlyingBird(Bird):

    def move(self):
        print("Flying in the sky.")

    def sound(self):
        print("Chirping.")

# 不可飛行的鳥類
class NonFlyingBird(Bird):
    def move(self):
        print("Running on the ground.")

    def sound(self):
        print("Making low noises.")

# 具體鳥類
class Sparrow(FlyingBird):
    def get_name(self):
        return "Sparrow"

class Ostrich(NonFlyingBird):
    def get_name(self):
        return "Ostrich"
```

```python
# 客戶端代碼
def handle_bird(bird: Bird):
    print(f"Handling a {bird.get_name()}:")
    print("Bird is moving:")
    bird.move()
    print("Bird sound:")
    bird.sound()
    print()

# 測試
birds = [Sparrow(), Ostrich()]

for bird in birds:
    handle_bird(bird)
```

### 擴展設計
如果需要新增如企鵝 (Penguin) 的不可飛行鳥類，只需繼承 NonFlyingBird，無需修改現有邏輯。

```python
class Penguin(NonFlyingBird):
    def get_name(self):
        return "Penguin"


# 測試
penguin = Penguin()
handle_bird(penguin)
```

### 重構的優點
- 滿足 LSP：每個子類別的行為符合父類別的預期。
- 高擴展性：新增鳥類僅需繼承相關類別，避免修改已有代碼。
- 可維護性：行為清晰，減少潛在錯誤。
- 更豐富的行為支持：抽象方法可進一步擴展系統功能。
